```python
# Multithreading
# When to use multi-threading
# It is used for I/o bound tasks. spending more time waiting for I/O operations (eg: file operations)
# Concurrent execution
```

```python
import threading
import time

def print_numbers():
    for i in range(5):
        time.sleep(2)
        print(f"Numer: {i}")

def print_letter():
    for letter in 'abcde':
        time.sleep(2)
        print(f"Letter: {letter}")

t = time.time()
print_numbers()
print_letter()
finished_time = t- time.time()
print(finished_time)
```

```python
import threading
import time

def print_numbers():
    for i in range(5):
        time.sleep(2)
        print(f"Numer: {i}")

def print_letter():
    for letter in 'abcde':
        time.sleep(2)
        print(f"Letter: {letter}")

# Creating threads
t1 = threading.Thread(target=print_numbers)
t2 = threading.Thread(target=print_letter)


t = time.time()
# start threads
t1.start()
t2.start()

t1.join()
t2.join()

finished_time = t- time.time()
print(finished_time)
```
