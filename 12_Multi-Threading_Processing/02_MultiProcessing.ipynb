```python
# Multi processing allows you to create processes that run in parallel
# When to use, based on CPU bound tasks. Mathematical computation, data processing
# Parallel execution by using all the cores of the cpu
```

```python
import multiprocessing
import time

def square_numbers():
    for i in range(5):
        time.sleep(1)
        print(f"Square : {i**2}")

def cube_numbers():
    for i in range(5):
        time.sleep(1)
        print(f"Cube : {i**3}")

if __name__== '__main__':
    # Create two processes
    p1 = multiprocessing.Process(target=square_numbers())
    p2 = multiprocessing.Process(target=cube_numbers())
    t = time.time()
    p1.start()
    p2.start()

    p1.join()
    p2.join()

    finished_time = t - time.time()
    print(finished_time)
```
