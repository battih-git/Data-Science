# Handling Imbalanced Dataset
1. Upsampling
2. Downsampling

```python
import numpy as np
import pandas as pd
np.random.seed(123)

n_samples = 1000
class_0_ratio = 0.9
n_class_0  = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0, n_class_1
```

```python
# Create a dataframe with imbalanced data set
class_0 = pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0] * n_class_0

})

class_1 = pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_1),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_1),
    'target':[1] * n_class_1
    
})
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
# Upsampling 
df_minority = df[df['target'] == 1]
df_majority = df[df['target'] == 0]
```

```python
from sklearn.utils import resample
upsampled = resample(df_minority,replace=True,n_samples=len(df_majority), random_state=42)
```

```python
upsampled.shape
```

```python
df_minority.head()
```

```python
pd.concat([df_minority,upsampled]).reset_index()
```

```python
# Downsampling
downsampled = resample(df_majority, replace=False, n_samples=len(df_minority), random_state=42)
```

```python

downsampled
```

```python
df_downsampled = pd.concat([df_minority,downsampled]).reset_index()
```
