In Python's logging module, you can use multiple loggers for different parts of your application. This is especially useful when you want to log messages from various components or modules separately, with different configurations (such as different log levels, output formats, or destinations).

```python
import logging
# Create a logger for module 1
logger1 = logging.getLogger('Module_1')
logger1.setLevel(logging.DEBUG)

# Create a logger for module 2
logger2 = logging.getLogger('Module 2')
logger2.setLevel(logging.WARNING)

logging.basicConfig(
    level=logging.DEBUG,
    format = '%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
```

```python
## Log messages with different logger 
logger1.debug("This is a debug message for Module 1")
logger2.warning("This is a warning message for Module 2")
logger2.error("This is an error message for module 2.")
```
