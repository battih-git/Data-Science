What is Logging in Python?

Logging in Python refers to the practice of recording events, information, and errors that occur during the execution of a program. It helps in tracking the program's behavior, debugging, and understanding what’s happening within the application, especially in production environments.

Python’s logging module provides a flexible framework for recording logs with various levels of severity (e.g., debug, info, warning, error, critical). This allows you to capture everything from detailed debugging information to high-level error reports.

# Python Logging

Logging is a crucial aspect of any application, providing a way to track events, errors, and operational information. Python's built-in logging module offers a flexible framework for emitting log messages from Python programs. In this lesson, we will cover the basics of logging, including how to configure logging, log levels, and best practices for using logging in Python applications.


```python
import logging

# Configure the basic logging settings
logging.basicConfig(level=logging.DEBUG)

## log messages
logging.debug("This is a debug message.")
logging.info("This is an info message.")
logging.warning("This is a warning message.")
logging.error("This is an error message.")
logging.critical('This is a critical message.')
```

```python
import logging
# Configuring logging
logging.basicConfig(
    filename='app.log',
    filemode='w',
    level = logging.INFO,
    format='%(asctime)s%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'  
)
```

```python
logger = logging.getLogger()
handler = logging.StreamHandler()
formatter = logging.Formatter('%(asctime)s - %(message)s', datefmt='%Y-%m-%d %H:%M:%S')
handler.setFormatter(formatter)
logger.addHandler(handler)
logger.setLevel(logging.INFO)
```
