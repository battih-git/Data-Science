# Data Visualization With Seaborn

Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code. In this lesson, we will cover the basics of Seaborn, including creating various types of plots and customizing them.


```python
import seaborn as sns
```

```python
# Basic Plotting with Seaborn
data = sns.load_dataset('tips')
data
```

```python
## Create a scatterplot
import matplotlib.pyplot as plt
%matplotlib inline
sns.scatterplot(x='total_bill',y='tip',data=data)
plt.title('Total Bill v/s Tips')
plt.show()
```

```python
# Line plot
sns.lineplot(x='size',y='total_bill', data=data)
plt.title('Line of Total Bill based on Size')
plt.show()
```

```python
# Categorical plot
sns.barplot(x='day',y='total_bill',data=data)
plt.title("Bar Plot of Total Bill by Day")
plt.show()
```

```python
# Box Plot
sns.boxplot(x='day',y='total_bill', data=data)
plt.title("Total Bill by Day")
plt.show()
```

```python
# Violin Plot 
sns.violinplot(x='day',y='total_bill',data=data)
```

```python
# Histogram Plot
sns.histplot(x='total_bill',bins=10,kde=True,data=data)
```

```python
## KDE Plot
sns.kdeplot(data['total_bill'],shade=True)
plt.show()
```

```python
# Pairplot
sns.pairplot(data=data,)
```

```python
# Heatmap
corr = data[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True)
```
