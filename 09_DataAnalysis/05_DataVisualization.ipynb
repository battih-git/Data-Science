# Data Visualization With Matplotlib

Matplotlib is a powerful plotting library for Python that enables the creation of static, animated, and interactive visualizations. It is widely used for data visualization in data science and analytics. In this lesson, we will cover the basics of Matplotlib, including creating various types of plots and customizing them.


```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
x = [1,2,3,4,5]
y = [1,4,9,16,25]
## Create a line plot
plt.plot(x,y)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Basic Line Plot")
plt.show()
```

```python
x = [1,2,3,4,5]
y = [1,4,9,16,25]

# Create a customized plot
plt.plot(x,y, color='blue', linestyle=':',marker = 'o',linewidth=1)
plt.grid()
plt.show()
```

```python
## Multiple Plots
## Sample data
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [1, 2, 3, 4, 5]
z = [2,4,6,8,10]

plt.figure(figsize=(9,5))
plt.subplot(2,2,1)
plt.plot(x,y1,color ='green')
plt.title("Plot 1")

plt.subplot(2,2,2)
plt.plot(x,y2,color ='red')
plt.title("Plot 2")

plt.subplot(2,2,3)
plt.plot(y1,y2,color ='blue')
plt.title("Plot 3")

plt.subplot(2,2,4)
plt.plot(z,y2,color ='yellow')
plt.title("Plot 4")

plt.show()
```

```python
# Bar Plot
categories = ['A','B','C','D','E']
values = [5,7,3,4,6]

# Create a bar plot
plt.bar(categories,values,color='green')
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Bar Plot")
plt.show()
```

### Histograms

Histograms are used to represent the distribution of a dataset. They divide the data into bins and count the number of data points in each bin.


```python
# Sample data
import numpy as np
data = np.random.normal(1,100,1000)
plt.hist(data, bins=5, color='orange',edgecolor = 'black')
```

### Scatter Plot

```python
x = [1,2,3,4,5]
y = [2,3,4,5,6]
plt.scatter(x,y,color='blue',)
```

### Pie Chart

```python
labels = ['A','B','C','D']
sizes = [30,20,40,10]
colors = ['gold','yellowgreen','lightcoral','lightskyblue']
explode = [0.1,0,0.1,0]
# Create a pie chart
plt.pie(sizes,labels = labels,colors=colors,autopct="%1.1f%%",explode=explode, shadow=True)
plt.title("Pie Chart")
plt.show()
```

```python
# Sales data visualization
import pandas as pd
df = pd.read_csv("Online Sales Data.csv")
df.head()
```

```python
total_sales_product = df.groupby('Product Category')['Total Revenue'].sum()
total_sales_product.plot(kind='bar',color='teal')
plt.show()
```

```python
# Plot sales trend over time
sales_trend = df.groupby('Date')['Total Revenue'].sum().reset_index()
sales_trend
```

```python
plt.plot(sales_trend['Date'],sales_trend['Total Revenue'])
```
