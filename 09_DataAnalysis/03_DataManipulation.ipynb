# Data Manipulation and Analysis with Pandas
Data manipulation and analysis are key tasks in any data science or data analysis project. Pandas provides a wide range of functions for data manipulation and analysis, making it easier to clean, transform, and extract insights from data. In this lesson, we will cover various data manipulation and analysis techniques using Pandas.

```python
import pandas as pd
df = pd.read_csv('data.csv') # Read the csv file
```

```python
# Print the first 5 rows
df.head()
```

```python
df.describe() # Statistical information about the data
```

```python
df.dtypes # To check the data type of the columns
```

```python
# Handling missing values
df.isnull().any()
```

```python
df.isnull().sum() #Checking the values where there are missing values
```

```python
# Filling missing values
df_fill = df.fillna(0)
df_fill.isnull().sum()
```

```python
# Indentifying a mean value for a specific column
df['Sales'].mean()
```

```python
# Filling missing values with the menan of the column
df['Sales'].fillna(df['Sales'].mean(),inplace=True)
df.head()
```

```python
df.dtypes
```

```python
# Rename columns
df.rename(columns={'Date':'Sales Date'},inplace=True)
df.head()
```

```python
# Chnage the data type of columns
df['Value'].fillna(df['Value'].mean(), inplace=True)
df['Value_new']  = df['Value'].astype(int)
df.head()
```

```python
df['New Value'] = df['Value'].apply(lambda x: x*2)
df.head()
```

```python
# Data Aggregating and Geouping
df.groupby('Product')['Value'].mean()
```

```python
# Using groupby to evaluate mean value at product and region level
df.groupby(['Product','Region'])['Value'].mean()
```

```python
# Using groupby to evaluate cumulative value at product and region level
df.groupby(['Product','Region'])['Value'].sum()
```

```python
# Using groupby to evaluate cumulative value at product and region level
df.groupby(['Product','Region'])['Value'].agg(['sum','mean','count'])
```

```python
# Merging and joining data frame
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
pd.merge(df1,df2,how='inner',on='Key')
```

```python
pd.merge(df1,df2, on='Key',how='outer')
```

```python
pd.merge(df1,df2, on='Key',how='left')
```

```python
pd.merge(df1,df2, on='Key',how='right')
```

```python
pd.merge(df1,df2, how='cross')
```
