# Decorators
Decorators are one of Pythons most powerful and elegant features, enabling you to modify or extend the behavior of functions or methods without changing their actual code. They allow you to wrap a function with additional functionality in a clean, readable way.

A decorator is a function that takes another function as an argument and adds functionality to it. It wraps the original function, modifying its behavior.

* function copy
* closures
* decorators 

```python
def welcome():
    return "Welcome to the advanced python course."
welcome()
```

```python
# Function copy: In Python, functions are objects, so we can assign them to new variables (like copying).
wel = welcome
print(welcome)
print(wel)
wel()
del welcome
wel()
```

```python
# Closures: A closure is a function that remembers values from its enclosing scope even after the outer function has finished executing.
def main_welcome(msg):
    def sub_welcome():
        print(msg)
        print('Welcome to advance python course.')
        print("Please learn these concepts properly.")
    
    return sub_welcome()

main_welcome("Welcome to Python")
```

```python
def main_welcome(func):

    def sub_welcome():
        print('Welcome to advance python course.')
        func("Welcome everyone to this tutorial")
        print("Please learn these concepts properly.")

    return sub_welcome()

main_welcome(print)
```

```python
def main_welcome(func):

    def sub_welcome():
        print('Welcome to advance python course.')
        print(func([1,2,3,4,5,6,7]))
        print("Please learn these concepts properly.")

    return sub_welcome()

main_welcome(len)
```

```python
# Decorator
def main_welcome(func):
    def sub_welcome_method():
        print("Welcome to advance python course")
        func()
        print("Please learn these concepts properly")
    return sub_welcome_method()
```

```python
def course_introduction():
    print("This is an advanced python course")

course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome
def course_introduction():
    print("This is an advanced python course")
```

```python
# Decorator

def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happending after the function is called.")
    return wrapper()

@my_decorator
def say_hello():
    print("Hello!")
```

```python
## Decorators with argument
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator

@repeat(3)
def say_hello():
    print("Hello")
```

```python
say_hello()
```
