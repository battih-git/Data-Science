# Generators
Generators are very simpler way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on file and do not store them in memory.

```python
def square(n):
    for i in range(n):
        return i**2
```

```python
square(3)
```

```python
def square(n):
    for i in range(n):
        yield i**2
```

```python
for i in square(3):
    print(i)
```

```python
a = square(3)
next(a)
```

```python
next(a)
```

```python
next(a)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
generator = my_generator()
next(generator)
```

```python
next(generator)
```

```python
next(generator)
```

```python
next(generator)
```

```python
gen = my_generator()
for value in gen:
    print(value)
```

```python
# Practical Example: Reading large files.

def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line


file_path = r"C:\Users\battih\Downloads\9dd7cd6e-eae8-45b3-a55d-117cefe67db5.csv"
for line in read_large_file(file_path):
    print(line.strip())
```
