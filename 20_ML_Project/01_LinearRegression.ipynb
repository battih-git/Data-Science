# Simple Linear Regression

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the dataset
df = pd.read_csv('height-weight(1).csv')
df.head()
```

```python
# Scatter plot
plt.scatter(df['Weight'],df['Height'])
plt.xlabel('Weight')
plt.ylabel('Height')
plt.title("Height v/s Weight")
```

```python
# Divide our dataset into dependent and independent features
X = df[['Weight']]
y = df['Height']
```

```python
# Split the data into train test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
# Standardize the dataset
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train_scaled,y_train)
y_pred = model.predict(X_test_scaled)
```

```python
print(f"The slope or coefficient of weight is: {model.coef_}")
```

```python
plt.scatter(X_train,y_train)
plt.plot(X_train,model.predict(X_train_scaled))
plt.plot(X_test,y_pred)
```

## Performance Metrics

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

mse = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
rmse = np.sqrt(mse)
print(mse)
print(mae)
print(rmse)
```

```python
score = r2_score(y_test,y_pred)
print(score)
```
