```python
# See the correlation 
df.corr()
```

```python
# View the correlation through chart
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(),annot=True)
```

```python
# Visualization
df['quality'].value_counts()
```

```python
# The dataset is an imbalanced dataset. The output categories are not equal.
df['quality'].value_counts().plot(kind='bar')
plt.ylabel('Count')
plt.xlabel('Wine Quality')
plt.title('Observations in Type of Wine Quality')
plt.show()
```

```python
# Distribution of each column
for column in df.columns:
    sns.histplot(df[column],kde=True)
```

```python
sns.histplot(df['alcohol'],kde=True)
```

```python
# Univariate, Bivariate and Multivariate analysis
sns.pairplot(df,hue='quality')
```

```python
# Categorical Plot
sns.catplot(x='quality',y='alcohol',data=df,kind='box')
```

```python
sns.scatterplot(data=df,x='density',y='pH',hue='quality') 
```
