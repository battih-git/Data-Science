# Abstraction
In Python, abstraction is typically achieved using abstract classes and abstract methods, which are defined in the abc module (Abstract Base Class). An abstract class can have abstract methods that must be implemented by subclasses, but the class itself doesn’t provide the full implementation.

In programming, abstraction means we hide the complicated stuff (the engine) and show you only what’s necessary (the controls you can interact with, like the steering wheel or gas pedal).

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
    
    @abstractmethod
    def perimeter(self):
        pass

# In this example, Shape is an abstract class that defines the interface for any shape. However, it doesn’t define the actual implementation of the area and perimeter methods.
```

```python
class Circle(Shape):
    def __init__(self, radius):
        super().__init__()
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius
    
    def perimeter(self):
        return 2 * 3.14 * self.radius
    
# Here, Circle and Rectangle are concrete classes that implement the abstract methods (area and perimeter). These classes provide the actual implementation details for each shape, 
# hiding those details from the user
```
