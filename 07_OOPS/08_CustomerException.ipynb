# Custom Exception (Raise and Throw Exception)

```python
# Adding custom exception for DOB if it stands invalid

class Error(Exception):
    pass
class dobException(Error):
    pass
```

```python
year = int(input("Enter the date of birth: "))
age = 2025 - year
try:    
    if age <= 30 and age >= 20:
        print("The age is valid so you can apply for the exam.")
    else:
        raise dobException
except dobException:
    print("Sorry, your not eligible for the exam.")
```
