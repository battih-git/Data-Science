# Magic Methods
In Python, magic methods (also called dunder methods, short for "double underscore" methods) are special methods that you can define in your classes to customize the behavior of objects in certain situations. These methods are used to implement operations like addition, string representation, comparison, etc.

They’re "magical" because Python calls them automatically when you perform certain actions on an object.

1. __init\__\(self, ...) — Object Constructor:

Purpose: This is the initializer or constructor method. It’s called when you create a new instance of a class (like when you create a new object).

Example: When you create a new object of a class, __init__ is called automatically to initialize the object’s attributes

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

dog = Dog('Buddy', 3)
print(dog.name)
print(dog.age)
```

2. __str\__(self) — String Representation

    Purpose: This method defines how an object is represented as a string (when you print it or use str() on the object).

    Example: If you print an object, Python calls __str__ to return a string that describes the object.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

dog = Dog('Buddy', 3)
print("Currently it will print the memory location of an object.")
print(dog)
```

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"{self.name}, is {self.age} year old."
dog = Dog('Buddy', 3)
print(dog)
```

3. __repr\__(self) — Official String Representation

    Purpose: __repr__ is similar to __str__, but it’s meant to provide a more detailed, unambiguous string representation. It's used by repr() and in debugging contexts like when you open a Python shell or print a list of objects.

    Example: It's often used for debugging to give more detail about the object.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __repr__(self):
        return f"Dog(name={self.name},age={self.age})"
dog = Dog('Buddy', 3)
print(repr(dog))
```
