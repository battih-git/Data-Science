# SQL And SQLite

SQL (Structured Query Language) is a standard language for managing and manipulating relational databases. SQLite is a lightweight, serverless, self-contained SQL database engine. It is widely used in mobile applications, desktop software, and small-scale web applications. Unlike other database systems (like MySQL or PostgreSQL), SQLite does not require a separate server process. Instead, it stores the entire database as a single file on the disk.

```python
import sqlite3
```

```python
## Connect to an SQLite database
connnection = sqlite3.connect("example.db")
connnection
```

```python
# Cursor is an object used to interact with the database. It is used to run query and 
cursor = connnection.cursor()
```

```python
# Create a table
cursor.execute('''
Create Table If Not Exists employees
               (
               id Integer Primary Key,
               name Text Not Null,
               age Integer,
               department text
               )
''')

## Commit the changes
connnection.commit()
```

```python
cursor.execute('''
select * from employees
''')
```

```python
# Insert a data in sqlite database
cursor.execute('''
Insert Into employees (name, age, department)
               Values('Bob', 25, 'Engineering')
''')

cursor.execute('''
Insert Into employees (name, age, department)
               Values('Charlie', 28, 'Finance')
''')

connnection.commit()
```

```python
# Query the data from the table
cursor.execute('Select * from employees')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
# Insert a value and fetch all records
cursor.execute('''
Insert into employees ('name','age','department')
Values ('Huzefa',31,'Finance')
               ''')
connnection.commit()

cursor.execute('Select * from employees')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
# Update the values in table
cursor.execute("""
Update employees
Set name = 'John', age=25
where id = 4
""")
connnection.commit()
```

```python
cursor.execute('Select * from employees')
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute(
    '''
Delete from employees 
where name = 'Bob'
'''
)
```

```python
cursor.execute('select * from employees')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
connnection.close()
```

```python
# Create a sales database
connnection = sqlite3.connect('sales_data.db')
cursor = connnection.cursor()
cursor.execute('''
create table if not exists sales(
               id integer primary key,
               date text not null,
               product text not null,
               sales integer,
               region text)
''')

sales_data = [
    ('2023-01-01', 'Product1', 100, 'North'),
    ('2023-02-01', 'Product2', 100, 'South'),
    ('2023-03-01', 'Product3', 100, 'East'),
    ('2023-04-01', 'Product4', 100, 'West'),
    ('2023-05-01', 'Product1', 100, 'North'),
    ('2023-06-01', 'Product2', 100, 'East'),
    ('2023-07-01', 'Product3', 100, 'West'),
]

cursor.executemany('''
Insert into sales (date, product, sales, region)
                   values(?,?,?,?)
                   ''',sales_data)
connnection.commit()
```

```python
cursor.execute('''select * from sales''')
rows = cursor.fetchall()
for row in rows:
    print(row)
```
