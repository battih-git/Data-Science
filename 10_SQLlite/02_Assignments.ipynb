1. Write a Python function to create a new SQLite3 database named `test.db`.

2. Write a Python function to create a table named `employees` with columns `id` (integer), `name` (text), `age` (integer), and `department` (text) in the `test.db` database.

```python
def create_table(name):
    connection = sqlite3.connect(name)
    cursor = connection.cursor()
    cursor.execute("""
    Create table if not exists employees(
                    id integer primary key,
                    name text not null,
                    age integer,
                    department text
                    )
    """)
    connection.commit()
    connection.close()
create_table('test.db')
```

```python
def insert_employee(name,age,department, database_name = 'test.db'):
    connection = sqlite3.connect(database_name,timeout=10)
    cursor = connection.cursor()
    cursor.executemany('''
    Insert into employees (name, age, department)
                   values(?,?,?)
    ''',[(name,age,department)])
    connection.commit()
    connection.close()
```

### Assignment 7: Handling Transactions

1. Write a Python function to insert multiple employees into the `employees` table in a single transaction. Ensure that if any insertion fails, none of the insertions are committed.

```python
def insert_multiple_employees(employees):
    connection = sqlite3.connect('test.db')
    cursor = connection.cursor()
    try:
        cursor.executemany(
            '''
        Insert Into employees(id,name,age,department)
        values (?,?,?,?)
            ''',
                employees)
        connection.commit()
        print("All employees data inserted successfully")
    except Exception as e:
        connection.rollback()
        print("Error occured, transaction rolled back.")
        print(e)
    finally:
        connection.close()

employees = [
    (6, 'Frank', 40, 'Finance'),
    (7, 'Grace', 29, 'Engineering'),
    (8, 'Hannah', 35, 'Marketing'),
    (9, 'Ivan', 38, 'Sales'),
    (6, 'Jack', 45, 'HR')  # Duplicate ID to cause an error
]
insert_multiple_employees(employees)
```

2. Write a Python function to update the age of multiple employees in a single transaction. Ensure that if any update fails, none of the updates are committed.

```python
def update_age(details):
    connection = sqlite3.connect('test.db')
    cursor = connection.cursor()
    try:
        cursor.executemany(
            '''
        Update employees
        set age = ?
        where id = ?
            ''',
                details)
        connection.commit()
        print("All employees data inserted successfully")
    except Exception as e:
        connection.rollback()
        print("Error occured, transaction rolled back.")
        print(e)
    finally:
        connection.close()

details = [
    (40,6),
    (29,7),
    (35,8),
    (20,9),
]
update_age(details)
fetch_database()
```

### Assignment 8: Creating Relationships

1. Create a new table named `departments` with columns `id` (integer) and `name` (text).

```python
def create_table(table_name):
    connection = sqlite3.connect('test.db')
    cursor = connection.cursor()
    cursor.execute(
        f'''
    Create table if not exists {table_name}(
        id integer primary key,
        department text
        )
    '''
        )
    connection.commit()
    connection.close()
create_table('department')
```

2. Modify the `employees` table to include a foreign key referencing the `id` column in the `departments` table

```python
import sqlite3

def add_department_foreign_key():
    # Connect to your SQLite database
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    
    # Disable foreign key checks
    cursor.execute('PRAGMA foreign_keys=off;')
    
    # Start a transaction
    cursor.execute('BEGIN TRANSACTION;')
    
    # Rename the old table
    cursor.execute('ALTER TABLE employees RENAME TO old_employees;')
    
    # Create the new table with the foreign key
    cursor.execute('''
    CREATE TABLE employees (
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        age INTEGER,
        department TEXT,
        department_id INTEGER,
        FOREIGN KEY (department_id) REFERENCES department(id)
    );
    ''')
    
    # Copy the data from the old table into the new table
    cursor.execute('''
    INSERT INTO employees (id, name, age, department, department_id)
    SELECT id, name, age, department, department_id
    FROM old_employees;
    ''')
    
    # Drop the old table
    cursor.execute('DROP TABLE old_employees;')
    
    # Commit changes and close the connection
    conn.commit()
    conn.close()

# Call the function to add the foreign key
add_department_foreign_key()
```
